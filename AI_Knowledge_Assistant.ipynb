```python
%pip install -q sentence-transformers gradio

import os
import torch
import gradio as gr
from sentence_transformers import SentenceTransformer, util
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pypdf import PdfReader

# Load AI embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Upload PDF
from google.colab import files
uploaded = files.upload()
pdf_name = next(iter(uploaded))

# Extract PDF text
reader = PdfReader(pdf_name)
pages_text = []

for page_no, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""
    if text.strip():
        pages_text.append({"page": page_no, "text": text})

# Split into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500, chunk_overlap=100
)

chunks = []
for page in pages_text:
    for chunk in splitter.split_text(page["text"]):
        chunks.append({"page": page["page"], "text": chunk})

chunk_texts = [c["text"] for c in chunks]
chunk_embeddings = model.encode(
    chunk_texts, convert_to_tensor=True
)

# Search PDF and return relevant answers
def ask_pdf(question):
    if not question.strip():
        return "Please enter a question."

    q_embedding = model.encode(
        question, convert_to_tensor=True
    )
    scores = util.cos_sim(q_embedding, chunk_embeddings)[0]
    top = scores.argsort(descending=True)[:3].tolist()

    results = []
    for i in top:
        results.append(
            f"Page {chunks[i]['page']} "
            f"(Similarity: {scores[i].item():.3f})\n"
            f"{chunks[i]['text']}"
        )

    return "\n\n--- Relevant source ---\n\n".join(results)

print("PDF:", pdf_name)
print("Pages extracted:", len(pages_text))
print("Chunks indexed:", len(chunks))
print("AI Knowledge Assistant is ready!")
```

```python
demo = gr.Interface(
    fn=ask_pdf,
    inputs=gr.Textbox(
        label="Ask a question about your PDF",
        placeholder="Example: What are the main objectives of this research?"
    ),
    outputs=gr.Textbox(label="Relevant answers from your document", lines=12),
    title="AI Knowledge Assistant",
    description="Ask questions and retrieve relevant information from your uploaded PDF."
)

demo.launch(share=True)
```
